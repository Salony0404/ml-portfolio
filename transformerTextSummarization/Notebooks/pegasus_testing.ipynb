```python
from transformers import PegasusForConditionalGeneration, PegasusTokenizer
```

```python
model_name = "google/pegasus-xsum"
# device = "cuda" if torch.cuda.is_available() else "cpu"
tokenizer = PegasusTokenizer.from_pretrained(model_name)
model = PegasusForConditionalGeneration.from_pretrained(model_name)
```

```python
sentence = "Hello My dog is cute"
tokenized_input = tokenizer(sentence)
# Print the tokenized input
print("Input IDs:", tokenized_input["input_ids"])
```

```python
# Decode the input tokens
decoded_input = tokenizer.decode(tokenized_input["input_ids"])
print("Decoded Input:", decoded_input)
```

```python
from transformers import PegasusTokenizer

# Load the Pegasus tokenizer
tokenizer = PegasusTokenizer.from_pretrained('google/pegasus-xsum')

# Example text to tokenize
example_text = "This is an example sentence."

# Tokenize the text
tokenized_output = tokenizer(example_text)

# Print the original text
print("Original Text:", example_text)

# Print the tokens formed by the tokenizer
print("Tokens:", tokenizer.convert_ids_to_tokens(tokenized_output['input_ids']))
```
