```python
import re

def remove_special_characters(text):
    # Check if the input is a string
    if isinstance(text, str):
        # Remove backticks, double asterisks, and hyphens
        text = text.replace("`", "")
        text = text.replace("**", "")
        text = text.replace("-", "")
        # Remove any remaining special characters
        text = re.sub(r'[^\w\s]', '', text)
    return text
```

```python
dataset["Explanation"] = dataset["Explanation"].apply(remove_special_characters)
```

```python
def remove_newlines1(text):
    # Check if the input is a string
    if isinstance(text, str):
        # Remove all occurrences of newline characters
        cleaned_text = text.replace("\n", "").replace("\r", "")
        return cleaned_text
    else:
        return text

```

```python
dataset["Explanation"] = dataset["Explanation"].apply(remove_newlines1)
```
